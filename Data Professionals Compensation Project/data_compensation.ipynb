```python
#Filling null with empty string. Usefull for concatenation
df["other_stack"]= df["other_stack"].fillna("")
```

```python
import pandas as pd

# Start with the gross salary column
salary = df["gross_salary"].astype("string").str.strip()

# Create a clean copy
df["gross_salary_clean"] = salary

# --------------------------------------------------
# 1. Convert USD to KES
# $1 = KES 130
# --------------------------------------------------

usd_mask = salary.str.contains(r"\$|usd", case=False, na=False)
hourly_mask = salary.str.contains(r"per hr", case=False, na=False)

usd_values = (
    salary[usd_mask & ~hourly_mask]
    .str.extract(r"([\d,]+(?:\.\d+)?)", expand=False)
    .str.replace(",", "", regex=False)
)

df.loc[usd_mask & ~hourly_mask, "gross_salary_clean"] = (
    usd_values.astype(float).mul(130).astype(str)
)


# --------------------------------------------------
# 2. Convert EUR to KES
# €1 = KES 150
# --------------------------------------------------

euro_mask = salary.str.contains("euro", case=False, na=False)

euro_values = (
    salary[euro_mask]
    .str.extract(r"([\d,]+(?:\.\d+)?)", expand=False)
    .str.replace(",", "", regex=False)
)

df.loc[euro_mask, "gross_salary_clean"] = (
    euro_values.astype(float).mul(150).astype(str)
)


# --------------------------------------------------
# 3. Convert "2.5 million"
# --------------------------------------------------

million_mask = salary.str.contains("million", case=False, na=False)

million_values = (
    salary[million_mask]
    .str.extract(r"([\d.]+)", expand=False)
)

df.loc[million_mask, "gross_salary_clean"] = (
    million_values.astype(float).mul(1_000_000).astype(str)
)


# --------------------------------------------------
# 4. Remove KES/Ksh formatting and commas
# --------------------------------------------------

df["gross_salary_clean"] = (
    df["gross_salary_clean"]
    .str.replace(",", "", regex=False)
    .str.replace("/-", "", regex=False)
    .str.replace("kshs", "", case=False, regex=False)
    .str.replace("ksh", "", case=False, regex=False)
    .str.replace("kes", "", case=False, regex=False)
    .str.strip()
)


# --------------------------------------------------
# 5. Convert K notation
# 80K -> 80000
# 120K -> 120000
# --------------------------------------------------

k_mask = df["gross_salary_clean"].str.match(
    r"^\d+(?:\.\d+)?k$",
    case=False,
    na=False
)

k_values = (
    df.loc[k_mask, "gross_salary_clean"]
    .str.replace("k", "", case=False, regex=False)
)

df.loc[k_mask, "gross_salary_clean"] = (
    k_values.astype(float).mul(1000).astype(str)
)


# --------------------------------------------------
# 6. Set hourly salaries to NULL
# --------------------------------------------------

df.loc[hourly_mask, "gross_salary_clean"] = pd.NA


# --------------------------------------------------
# 7. Set non-salary responses to NULL
# --------------------------------------------------

missing_values = [
    "nan",
    "Nan",
    "Nill",
    "Nil",
    "Unemployed",
    "Prefer not to say"
]

df["gross_salary_clean"] = df["gross_salary_clean"].replace(
    missing_values,
    pd.NA
)


# --------------------------------------------------
# 8. Convert to numeric
# --------------------------------------------------

df["gross_salary_clean"] = pd.to_numeric(
    df["gross_salary_clean"],
    errors="coerce"
)


# --------------------------------------------------
# 9. Remove zero/negative salaries
# --------------------------------------------------

df.loc[
    df["gross_salary_clean"] <= 0,
    "gross_salary_clean"
] = pd.NA


# --------------------------------------------------
# 10. Convert to nullable integer
# --------------------------------------------------

df["gross_salary_clean"] = df["gross_salary_clean"].astype("Int64")
df["gross_salary"]= df["gross_salary_clean"]
df["gross_salary"].unique()
```

```python
#Work setup cleaning
df["work_setup"]= df["work_setup"].replace({
    'Hybrid- I need to be in the office specific days in a week': 'Hybrid- Specific office days',
    'Hybrid-I choose my days to go the office': "Hybrid- Choose office days"
})
df["work_setup"].unique()
```
